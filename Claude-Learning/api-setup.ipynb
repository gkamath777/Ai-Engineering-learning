#### Create An API Client

```python
from anthropic import Anthropic

client = Anthropic()
model = "claude-haiku-4-5-20251001"
```

#### Make a request  

```python
message = client.messages.create (
    model = model,
    max_tokens= 1000,
    messages = [
        {
            "role":"user",
            "content": "whats is Quantum Computing"
        }
    ]
)
message
print(message.content[0].text)
```

```python
def add_user_message(messages, text): 
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create (
        model = model,
        max_tokens= 1000,
        messages = messages
    )
    return message.content[0].text
```

```python
# Mak a starting of list of messages
messages= []

# add a initial user question
add_user_message(messages, "Define Quantum computing")

print(messages)
print("=======================")

#Pass the list of messages to the chat to get the answer
answer = chat(messages)
print(f"Answer: \n{answer}")
print("=======================")

# Add the previous messages to the assisstant
add_assistant_message(messages, answer)
print(messages)
print("="*len(messages))

# Add in the user;s another followup questions
add_user_message(messages, "Write anoter statement")

# Call chat again with the list of messages
answer_new = chat(messages)
print(f"Answer: \n{answer_new}")
```

```python
# Define the message object
messages = []

# Use "While True" loop to run the chatbot forever
flag = True
while flag:
    #get user input
    user_input = input("> ")
    print(f"> {user_input}")
    
    if user_input == "exit":
        flag=False
        break

    if user_input == "new":
        messages=[]
        continue

    # Add usr input to the list of messages
    add_user_message(messages, user_input)

    # Call claude with the 'chat' function
    answer = chat(messages)

    # Add generated text to the list of messages
    add_assistant_message(messages, answer)

    # Print the generated text
    print(f"Answer: \n{answer}")

```
