```python
from dotenv import load_dotenv
from anthropic import Anthropic


load_dotenv()
client = Anthropic()
model = "claude-haiku-4-5-20251001"

SYSTEM_PROMPT = """
you are a patient math tutor. Do not directly answer a student's questions. Guide them to a solution step by step
"""

def add_user_message(messages, text): 
    user_message = {"role": "user", "content": text}
    messages.append(user_message)


def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)


def chat(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages
    }

    if system:
        params["system"] = system

    message = client.messages.create (**params)
    return message.content[0].text
```

```python
#Calculator chatbot that takes user input and performs calculations using Claude API
# Define the message object
messages = []
answer = '0'

# Use "While True" loop to run the chatbot forever
flag = True
while flag:
    #get user input
    print(f"current value {answer}\n")
    print("add the number(s) and mention the operations to perform \n")
    print("for New Chat type 'new' and to exit type 'exit'")
    user_input = input("> ")
    print(f"> {user_input}")
    
    if user_input == "exit":
        flag=False
        break

    if user_input == "new":
        messages=[]
        continue

    # Add usr input to the list of messages
    add_user_message(messages, user_input)

    # Call claude with the 'chat' function
    answer = chat(messages)

    # Add generated text to the list of messages
    add_assistant_message(messages, answer)

    # Print the generated text
    print(f"Answer: \n{answer}")
```

```python
messages = []
add_user_message(messages, "who is the experience scientist in Quantam Physics")

message = client.messages.create (
        model = model,
        max_tokens= 1000,
        messages = messages,
        stream = True
)


for event in message:
    print(event)
```
